# Example: Running DplPy with Jupyter Notebook

This Jupyter Notebook must be run within the `dplPy` directory. Refer to `README.md` for instructions on how to run Jupyter Notebook and the required `conda Environment` on your OS.

Import packages and load test data from the `test` directory.
> **Note**: choose your own example file from `/tests/data/<format>/<file.format>`. In this example, the test file is `rwl/ca533.rwl`. 

```python
# Import dplPy and load an example dataset.
# (dplPy installs via pip; the example data live in the repo's tests/ folder,
#  referenced here relative to this notebook.)
import dplpy as dpl

data = dpl.readers("../tests/data/rwl/ca533.rwl")
```

Summarize data using the `summary` function.

```python
# Summarizes input data

dpl.summary(data)
```

Show general statistics of `data` through the `stats` function.

```python
# General statistics of input data

dpl.stats(data)
```

```python
# Detrend a single series. `fit` chooses the growth CURVE and `method` chooses
# the ARITHMETIC: method="ratio" (default) divides the series by the curve;
# method="difference" subtracts it. Available fits include "Spline" (default),
# "AgeDepSpline", "ModNegExp", "ModHugershoff", "Hugershoff", "Linear", "Mean".
# detrend() accepts a single series or a whole dataframe:

dpl.detrend(data["CAM191"])
```

```python
# Returns parameters
# As default max lag is set to 5; 
# add second parameter (integer) to change max lag.

dpl.autoreg(data["CAM191"], 10)
```

```python
# Returns residuals+mean after 
# choosing best AR model fit with chosen max lag (default=5).

dpl.ar_func(data["CAM191"])
```
